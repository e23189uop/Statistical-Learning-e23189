#Q1) Prediction of fraudulent transactions

A bank monitors credit-card transactions for possible fraud.

For each transaction, let the sample space be

$$
\Omega=\{\omega_1,\omega_2,\omega_3,\omega_4\},
$$

where:

* $\omega_1$: low-risk customer, legitimate transaction,
* $\omega_2$: low-risk customer, fraudulent transaction,
* $\omega_3$: high-risk customer, legitimate transaction,
* $\omega_4$: high-risk customer, fraudulent transaction.

Assume the probabilities are

$$
P(\{\omega_1\})=0.50,\qquad
P(\{\omega_2\})=0.10,\qquad
P(\{\omega_3\})=0.20,\qquad
P(\{\omega_4\})=0.20.
$$

Let $X:\Omega\to\mathbb R$ denote the fraud indicator:

$$
X(\omega_1)=0,\qquad
X(\omega_2)=1,\qquad
X(\omega_3)=0,\qquad
X(\omega_4)=1.
$$

Thus $X=1$ means the transaction is fraudulent, and $X=0$ means it is legitimate.

Now suppose the bank does not observe the full outcome $\omega$, but only the customer's risk class. Define $Y:\Omega\to\mathbb R$ by

$$
Y(\omega_1)=0,\qquad
Y(\omega_2)=0,\qquad
Y(\omega_3)=1,\qquad
Y(\omega_4)=1.
$$

Thus:

* $Y=0$ means “low-risk customer,”
* $Y=1$ means “high-risk customer.”

Answer the following.

###Answer:


## 1. The Probability Space $(\Omega, \mathfrak{F}, P)$

A formal probability space requires a sample space, a collection of measurable events, and a probability measure:


**Sample Space ($\Omega$):** $\Omega = \{\omega_1, \omega_2, \omega_3, \omega_4\}$.



**$\sigma$-algebra ($\mathfrak{F}$):** Since $\Omega$ is a finite set, we use the power set $\mathfrak{F} = \mathfrak{P}(\Omega)$. This contains all $2^4 = 16$ possible subsets of $\Omega$, ensuring every subset is a measurable event.



**Probability Measure ($P$):** Defined on the elementary singletons:



$$P(\{\omega_1\}) = 0.50, \quad P(\{\omega_2\}) = 0.10, \quad P(\{\omega_3\}) = 0.20, \quad P(\{\omega_4\}) = 0.20$$


This is a valid probability measure because $\sum_{i=1}^4 P(\{\omega_i\}) = 0.50 + 0.10 + 0.20 + 0.20 = 1$.



---

## 2. The $\sigma$-algebra Generated by $Y$, namely $\sigma(Y)$

The random variable $Y$ serves as our information filter. To find $\sigma(Y)$, we determine the preimages (inverse images) of all possible values in the range of $Y$:

 For $Y = 0$: $Y^{-1}(\{0\}) = \{\omega_1, \omega_2\}$


 For $Y = 1$: $Y^{-1}(\{1\}) = \{\omega_3, \omega_4\}$



The $\sigma$-algebra $\sigma(Y)$ is the smallest $\sigma$-algebra containing these preimages. It represents the collection of events we can "see" if we only observe $Y$:


$$\sigma(Y) = \{\emptyset, \Omega, \{\omega_1, \omega_2\}, \{\omega_3, \omega_4\}\}$$



---

## 3. Measurability and Constants on Atoms

By definition, the conditional expectation $\mathbb{E}[X \mid Y]$ is a random variable that is **$\sigma(Y)$-measurable**.

From a measure-theoretic standpoint, a random variable is measurable with respect to a sub-$\sigma$-algebra if it can be written as a deterministic function of the underlying generating variable (i.e., $\mathbb{E}[X \mid Y] = g(Y)$).

Because $\sigma(Y)$ cannot distinguish individual elements within its minimal non-empty sets (called **atoms**), any function measurable with respect to it must treat elements inside an atom identically . Here, the atoms are exactly $\{\omega_1, \omega_2\}$ and $\{\omega_3, \omega_4\}$ . Therefore, $\mathbb{E}[X \mid Y]$ cannot vary within those sets and **must be constant** across them.

We can express this structurally as:


$$\mathbb{E}[X \mid Y](\omega) = \begin{cases} a, & \omega \in \{\omega_1, \omega_2\} \\ b, & \omega \in \{\omega_3, \omega_4\} \end{cases}$$

---

## 4. Explicit Computation of $\mathbb{E}[X \mid Y]$

To find the constants $a$ and $b$, we invoke the defining property of conditional expectation: $\int_A \mathbb{E}[X \mid Y] \, dP = \int_A X \, dP$ for all $A \in \sigma(Y)$.

### On the set $A = \{\omega_1, \omega_2\}$:


**Left-Hand Side:** $\int_{\{\omega_1, \omega_2\}} a \, dP = a \cdot P(\{\omega_1, \omega_2\}) = a(0.50 + 0.10) = 0.60a$



**Right-Hand Side:** $\int_{\{\omega_1, \omega_2\}} X \, dP = X(\omega_1)P(\{\omega_1\}) + X(\omega_2)P(\{\omega_2\}) = 0(0.50) + 1(0.10) = 0.10$



**Equating both:** $0.60a = 0.10 \implies a = \frac{1}{6}$



### On the set $A = \{\omega_3, \omega_4\}$:


**Left-Hand Side:** $\int_{\{\omega_3, \omega_4\}} b \, dP = b \cdot P(\{\omega_3, \omega_4\}) = b(0.20 + 0.20) = 0.40b$



**Right-Hand Side:** $\int_{\{\omega_3, \omega_4\}} X \, dP = X(\omega_3)P(\{\omega_3\}) + X(\omega_4)P(\{\omega_4\}) = 0(0.20) + 1(0.20) = 0.20$



**Equating both:** $0.40b = 0.20 \implies b = \frac{1}{2}$



Putting this together using indicator functions yields the desired representation:


$$\mathbb{E}[X \mid Y] = \frac{1}{6} \mathbb{1}_{\{Y=0\}} + \frac{1}{2} \mathbb{1}_{\{Y=1\}}$$



---

## 5. Verifying the Defining Property

We verify $\int_A \mathbb{E}[X \mid Y] \, dP = \int_A X \, dP$ across all four elements of $\sigma(Y)$:


**$A = \emptyset$:** $0 = 0$ (Trivially true).



**$A = \{\omega_1, \omega_2\}$:** We already solved this above: $0.10 = 0.10$.



**$A = \{\omega_3, \omega_4\}$:** We already solved this above: $0.20 = 0.20$.



**$A = \Omega$:** * $\int_{\Omega} \mathbb{E}[X \mid Y] \, dP = \frac{1}{6}(0.60) + \frac{1}{2}(0.40) = 0.10 + 0.20 = 0.30$



$\int_{\Omega} X \, dP = 0(0.50) + 1(0.10) + 0(0.20) + 1(0.20) = 0.30$





Since it holds for $A = \Omega$, this beautifully checks out with the **Law of Total Expectation**: $\mathbb{E}[\mathbb{E}[X \mid Y]] = \mathbb{E}[X] = 0.30$.

---

## 6. Conditional Probability Mass Function $p_{X \mid Y}(x \mid y)$

Using the discrete definition $p_{X \mid Y}(x \mid y) = \frac{P(X=x, Y=y)}{P(Y=y)}$:


**Given $Y = 0$** ($P(Y=0) = 0.60$):



$p_{X \mid Y}(1 \mid 0) = \frac{P(\{\omega_2\})}{0.60} = \frac{0.10}{0.60} = \frac{1}{6}$



$p_{X \mid Y}(0 \mid 0) = \frac{0.50}{0.60} = \frac{5}{6}$





**Given $Y = 1$** ($P(Y=1) = 0.40$):



$p_{X \mid Y}(1 \mid 1) = \frac{P(\{\omega_4\})}{0.40} = \frac{0.20}{0.40} = \frac{1}{2}$



$p_{X \mid Y}(0 \mid 1) = \frac{0.20}{0.40} = \frac{1}{2}$





---

## 7. Verifying the Identity

Let's confirm that the evaluation matches our functional forms:


**For $y = 0$:** $\sum_{x \in \{0,1\}} x \cdot p_{X \mid Y}(x \mid 0) = 0\left(\frac{5}{6}\right) + 1\left(\frac{1}{6}\right) = \frac{1}{6}$. This matches $\mathbb{E}[X \mid Y=0]$.



**For $y = 1$:** $\sum_{x \in \{0,1\}} x \cdot p_{X \mid Y}(x \mid 1) = 0\left(\frac{1}{2}\right) + 1\left(\frac{1}{2}\right) = \frac{1}{2}$. This matches $\mathbb{E}[X \mid Y=1]$.



---

## 8. Mean Square Error (MSE)

The Minimum Mean Square Error (MMSE) for the optimal estimator $\mathbb{E}[X \mid Y]$ equals the expected conditional variance:


$$MSE_{min} = \mathbb{E}[\text{Var}(X \mid Y)]$$



Because $X$ given $Y=y$ functions as a Bernoulli variable, its variance is $p(1-p)$:


$\text{Var}(X \mid Y=0) = \frac{1}{6}\left(1 - \frac{1}{6}\right) = \frac{5}{36}$



$\text{Var}(X \mid Y=1) = \frac{1}{2}\left(1 - \frac{1}{2}\right) = \frac{1}{4}$



Now take the expectation over the probability distribution of $Y$:


$$MSE_{min} = \left(\frac{5}{36} \cdot 0.60\right) + \left(\frac{1}{4} \cdot 0.40\right) = \frac{1}{12} + \frac{1}{10} = \frac{11}{60} \approx 0.1833$$



---

## 9. Global Estimator vs. Law of Total Variance

If the bank drops the risk class $Y$ entirely and uses the flat global baseline $\mathbb{E}[X] = 0.30$, the error calculation defaults to the total variance of $X$:


$$MSE_{global} = \text{Var}(X) = p(1-p) = 0.30(1 - 0.30) = 0.21$$



### Relationship to the Law of Total Variance:

The Law of Total Variance asserts that:


$$\text{Var}(X) = \mathbb{E}[\text{Var}(X \mid Y)] + \text{Var}(\mathbb{E}[X \mid Y])$$



Let's compute the variance of our estimator ($\text{Var}(\mathbb{E}[X \mid Y])$) to see how much error was systematically eliminated by knowing $Y$:


$\mathbb{E}[(\mathbb{E}[X \mid Y])^2] = \left(\frac{1}{36} \cdot 0.60\right) + \left(\frac{1}{4} \cdot 0.40\right) = 0.01667 + 0.10 = \frac{7}{60} \approx 0.1167$



$\text{Var}(\mathbb{E}[X \mid Y]) = \mathbb{E}[(\mathbb{E}[X \mid Y])^2] - (\mathbb{E}[X])^2 = 0.1167 - (0.30)^2 = 0.0267$



Decomposition breakdown:


$$0.21 \text{ (Total Variance)} = 0.1833 \text{ (Unexplained Error)} + 0.0267 \text{ (Explained Variance Reduction)}$$



---

## 10. Information-Theoretic Analysis

### a) Entropy $H(X)$

This measures the bank's baseline uncertainty regarding transaction fraud:


$$H(X) = - (0.70 \log_2 0.70 + 0.30 \log_2 0.30) \approx 0.881 \text{ bits}$$



### b) Conditional Entropy $H(X \mid Y)$

This calculates the remaining uncertainty after $Y$ is revealed:


$H(X \mid Y=0) = - \left(\frac{1}{6} \log_2 \frac{1}{6} + \frac{5}{6} \log_2 \frac{5}{6}\right) \approx 0.650 \text{ bits}$



$H(X \mid Y=1) = - \left(\frac{1}{2} \log_2 \frac{1}{2} + \frac{1}{2} \log_2 \frac{1}{2}\right) = 1.00 \text{ bit}$



Averaging over the distribution of $Y$:


$$H(X \mid Y) = (0.60 \times 0.650) + (0.40 \times 1.00) = 0.390 + 0.40 = 0.790 \text{ bits}$$



### c) Mutual Information $I(X; Y)$

The true information reduction gained by processing the feature $Y$:


$$I(X; Y) = H(X) - H(X \mid Y) = 0.881 - 0.790 = 0.091 \text{ bits}$$



### d) Percentage of Uncertainty Removed

$$\text{Percentage} = \frac{I(X; Y)}{H(X)} = \frac{0.091}{0.881} \approx 10.3\%$$



By assessing the risk profile $Y$, the bank strips away exactly 10.3% of its overall uncertainty regarding fraud.

# Q2) Conditional Density of the Multivariate Gaussian

Let $\mathbf{Z}$ be the concatenated random matrix
$\mathbf{Z} = \begin{bmatrix} \mathbf{X} \\ \mathbf{Y} \end{bmatrix} \in \mathbb{R}^{n+m}$.


The random variable $\mathbf{Z}$ is said to have a **joint multivariate normal distribution** if every linear combination of its components follows a univariate normal distribution.

For $\mathbf{Z}$ to possess a **joint density function** (meaning its distribution is absolutely continuous with respect to the Lebesgue measure on $\mathbb{R}^{n+m}$), its covariance matrix must be non-singular (positive definite).


The joint density function $f_{\mathbf{X}, \mathbf{Y}}(\mathbf{z})$ for $\mathbf{z} \in \mathbb{R}^{n+m}$ is given by:

$$f_{\mathbf{X}, \mathbf{Y}}(\mathbf{z}) = \frac{1}{\sqrt{(2\pi)^{n+m} |\boldsymbol{\Sigma}|}} \exp\left( -\frac{1}{2} (\mathbf{z} - \boldsymbol{\mu})^T \boldsymbol{\Sigma}^{-1} (\mathbf{z} - \boldsymbol{\mu}) \right).$$

Here:

* **Mean ($\boldsymbol{\mu}$):** $\boldsymbol{\mu} = E[\mathbf{Z}] = \begin{bmatrix} \boldsymbol{\mu}_X \\ \boldsymbol{\mu}_Y \end{bmatrix} \in \mathbb{R}^{n+m}$, where $\boldsymbol{\mu}_X \in \mathbb{R}^n$ and $\boldsymbol{\mu}_Y \in \mathbb{R}^m$.
* **Covariance Matrix ($\boldsymbol{\Sigma}$):** A symmetric, positive definite matrix $\boldsymbol{\Sigma} \in \mathbb{R}^{(n+m) \times (n+m)}$ partitioned as:
$$\boldsymbol{\Sigma} = \begin{bmatrix} \boldsymbol{\Sigma}_{XX} & \boldsymbol{\Sigma}_{XY} \\ \boldsymbol{\Sigma}_{YX} & \boldsymbol{\Sigma}_{YY} \end{bmatrix}$$
    * $\boldsymbol{\Sigma}_{XX} = \text{Var}(\mathbf{X})$ (size $n \times n$)
    * $\boldsymbol{\Sigma}_{YY} = \text{Var}(\mathbf{Y})$ (size $m \times m$)
    * $\boldsymbol{\Sigma}_{XY} = \text{Cov}(\mathbf{X}, \mathbf{Y}) = \boldsymbol{\Sigma}_{YX}^T$ (size $n \times m$)
* **Determinant ($|\boldsymbol{\Sigma}|$):** The normalization factor ensuring the total integral is 1.

Show that the conditional distribution is given by $f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x} \mid \mathbf{y})$

$$f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x} \mid \mathbf{y}) = \frac{1}{\sqrt{(2\pi)^n |\mathbf{S}_{XX}|}} \exp\left( -\frac{1}{2} (\mathbf{x} - \boldsymbol{\mu}^*)^T \boldsymbol{\Sigma}_{X\mid Y}^{-1} (\mathbf{x} - \boldsymbol{\mu}^*) \right),$$
where
* $\boldsymbol{\Sigma}_{X\mid Y} = \boldsymbol{\Sigma}_{XX} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}$
* $\boldsymbol{\mu}^* = \boldsymbol{\mu}_x + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}(\mathbf{y} - \boldsymbol{\mu}_y)$

###Answer:

To derive the conditional distribution $f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x} \mid \mathbf{y})$ of a multivariate normal vector, we can use two primary methods: the **Linear Transformation (Geometric Projection) Method** and the **Direct Algebraic (Density Division) Method**.


---

### Method 1: The Linear Transformation Method (The Elegant Path)

Instead of grinding through heavy matrix algebra inside an exponent, this method uses a clever change of variables to decouple $\mathbf{X}$ and $\mathbf{Y}$.

#### Step 1: Construct an un-correlated random variable

We want to define a new random variable $\mathbf{W} \in \mathbb{R}^n$ as a linear combination of $\mathbf{X}$ and $\mathbf{Y}$:


$$\mathbf{W} = \mathbf{X} - \mathbf{A}\mathbf{Y}$$


We choose the matrix $\mathbf{A} \in \mathbb{R}^{n \times m}$ such that $\mathbf{W}$ and $\mathbf{Y}$ are **uncorrelated**, meaning $\text{Cov}(\mathbf{W}, \mathbf{Y}) = \mathbf{0}$:


$$\text{Cov}(\mathbf{W}, \mathbf{Y}) = \text{Cov}(\mathbf{X} - \mathbf{A}\mathbf{Y}, \mathbf{Y}) = \text{Cov}(\mathbf{X}, \mathbf{Y}) - \mathbf{A}\text{Var}(\mathbf{Y}) = \boldsymbol{\Sigma}_{XY} - \mathbf{A}\boldsymbol{\Sigma}_{YY}$$

Setting this to $\mathbf{0}$ yields:


$$\boldsymbol{\Sigma}_{XY} - \mathbf{A}\boldsymbol{\Sigma}_{YY} = \mathbf{0} \implies \mathbf{A} = \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}$$

#### Step 2: Establish Joint Distribution and Independence

Because $\mathbf{W}$ and $\mathbf{Y}$ are linear transformations of the jointly Gaussian vector $\mathbf{Z}$, they are also **jointly multivariate normal**.

For multivariate normal random variables, being **uncorrelated implies independence**. Therefore, $\mathbf{W}$ and $\mathbf{Y}$ are completely independent.

#### Step 3: Compute the Mean and Variance of $\mathbf{W}$

Since $\mathbf{W} = \mathbf{X} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\mathbf{Y}$, its properties are straightforward to calculate:

* **Mean:**

$$\boldsymbol{\mu}_W = \mathbb{E}[\mathbf{W}] = \mathbb{E}[\mathbf{X}] - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\mathbb{E}[\mathbf{Y}] = \boldsymbol{\mu}_X - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\mu}_Y$$


* **Covariance:**

$$\text{Var}(\mathbf{W}) = \text{Var}(\mathbf{X} - \mathbf{A}\mathbf{Y}) = \text{Var}(\mathbf{X}) - \mathbf{A}\text{Cov}(\mathbf{Y}, \mathbf{X}) - \text{Cov}(\mathbf{X}, \mathbf{Y})\mathbf{A}^T + \mathbf{A}\text{Var}(\mathbf{Y})\mathbf{A}^T$$



Substituting $\mathbf{A} = \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}$:

$$\text{Var}(\mathbf{W}) = \boldsymbol{\Sigma}_{XX} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX} + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}$$


$$\text{Var}(\mathbf{W}) = \boldsymbol{\Sigma}_{XX} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX} = \boldsymbol{\Sigma}_{X\mid Y}$$



Thus, $\mathbf{W} \sim \mathcal{N}\left(\boldsymbol{\mu}_X - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\mu}_Y, \, \boldsymbol{\Sigma}_{X\mid Y}\right)$.

#### Step 4: Condition on $\mathbf{Y} = \mathbf{y}$

We can rewrite our original variable $\mathbf{X}$ as:


$$\mathbf{X} = \mathbf{W} + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\mathbf{Y}$$

When we condition on the event $\mathbf{Y} = \mathbf{y}$:

1. The term $\boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\mathbf{Y}$ becomes a deterministic constant vector $\boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\mathbf{y}$.
2. Because $\mathbf{W}$ is independent of $\mathbf{Y}$, its distribution remains completely unaffected by the conditioning.

Thus, the conditional distribution of $\mathbf{X}$ given $\mathbf{Y}=\mathbf{y}$ is simply the distribution of $\mathbf{W}$ shifted by a constant:


$$\mathbf{X} \mid \mathbf{Y}=\mathbf{y} \sim \mathcal{N}\left(\mathbb{E}[\mathbf{W}] + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\mathbf{y}, \, \text{Var}(\mathbf{W})\right)$$

Substituting the mean and variance of $\mathbf{W}$:


$$\boldsymbol{\mu}^* = \left(\boldsymbol{\mu}_X - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\mu}_Y\right) + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\mathbf{y} = \boldsymbol{\mu}_X + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}(\mathbf{y} - \boldsymbol{\mu}_Y)$$

$$\text{Var}(\mathbf{X} \mid \mathbf{Y}=\mathbf{y}) = \boldsymbol{\Sigma}_{X\mid Y} = \boldsymbol{\Sigma}_{XX} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}$$

Since $\mathbf{X}\mid\mathbf{Y}=\mathbf{y}$ is a normal distribution with mean $\boldsymbol{\mu}^*$ and covariance $\boldsymbol{\Sigma}_{X\mid Y}$, its probability density function is explicitly:


$$f_{\mathbf{X}\mid\mathbf{Y}}(\mathbf{x} \mid \mathbf{y}) = \frac{1}{\sqrt{(2\pi)^n |\boldsymbol{\Sigma}_{X\mid Y}|}} \exp\left( -\frac{1}{2} (\mathbf{x} - \boldsymbol{\mu}^*)^T \boldsymbol{\Sigma}_{X\mid Y}^{-1} (\mathbf{x} - \boldsymbol{\mu}^*) \right)$$

---

### Method 2: Direct Algebraic Proof (Density Division)

If you need to show it directly from the joint density definition, we evaluate the definition of conditional density:


$$f_{\mathbf{X}\mid\mathbf{Y}}(\mathbf{x} \mid \mathbf{y}) = \frac{f_{\mathbf{X}, \mathbf{Y}}(\mathbf{x}, \mathbf{y})}{f_{\mathbf{Y}}(\mathbf{y})}$$

#### Step 1: Partitioning the Determinant (Schur Complement)

The determinant of the full covariance matrix can be factored using the Schur complement of $\boldsymbol{\Sigma}_{YY}$:


$$|\boldsymbol{\Sigma}| = |\boldsymbol{\Sigma}_{YY}| \cdot |\boldsymbol{\Sigma}_{XX} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}| = |\boldsymbol{\Sigma}_{YY}| \cdot |\boldsymbol{\Sigma}_{X\mid Y}|$$

When dividing the normalization factors of the joint density by the marginal density of $\mathbf{Y}$, we get:


$$\frac{\frac{1}{\sqrt{(2\pi)^{n+m}|\boldsymbol{\Sigma}|}}}{\frac{1}{\sqrt{(2\pi)^m|\boldsymbol{\Sigma}_{YY}|}}} = \frac{1}{(2\pi)^{n/2}} \sqrt{\frac{|\boldsymbol{\Sigma}_{YY}|}{|\boldsymbol{\Sigma}|}} = \frac{1}{\sqrt{(2\pi)^n |\boldsymbol{\Sigma}_{X\mid Y}|}}$$

#### Step 2: Partitioning the Inverse Covariance Matrix

Using the block matrix inversion formula, the inverse of $\boldsymbol{\Sigma}$ can be written in terms of $\boldsymbol{\Sigma}_{X\mid Y}$ and the shorthand matrix $\mathbf{A} = \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}$:


$$\boldsymbol{\Sigma}^{-1} = \begin{bmatrix} \boldsymbol{\Sigma}_{X\mid Y}^{-1} & -\boldsymbol{\Sigma}_{X\mid Y}^{-1}\mathbf{A} \\ -\mathbf{A}^T\boldsymbol{\Sigma}_{X\mid Y}^{-1} & \boldsymbol{\Sigma}_{YY}^{-1} + \mathbf{A}^T\boldsymbol{\Sigma}_{X\mid Y}^{-1}\mathbf{A} \end{bmatrix}$$

#### Step 3: Simplifying the Exponent Matrix Algebra

Let $\tilde{\mathbf{x}} = \mathbf{x} - \boldsymbol{\mu}_X$ and $\tilde{\mathbf{y}} = \mathbf{y} - \boldsymbol{\mu}_Y$. The quadratic term in the joint density exponent is:


$$Q_{\text{joint}} = \begin{bmatrix} \tilde{\mathbf{x}} \\ \tilde{\mathbf{y}} \end{bmatrix}^T \boldsymbol{\Sigma}^{-1} \begin{bmatrix} \tilde{\mathbf{x}} \\ \tilde{\mathbf{y}} \end{bmatrix}$$

Expanding this block multiplication yields:


$$Q_{\text{joint}} = \tilde{\mathbf{x}}^T \boldsymbol{\Sigma}_{X\mid Y}^{-1} \tilde{\mathbf{x}} - \tilde{\mathbf{x}}^T \boldsymbol{\Sigma}_{X\mid Y}^{-1}\mathbf{A}\tilde{\mathbf{y}} - \tilde{\mathbf{y}}^T \mathbf{A}^T \boldsymbol{\Sigma}_{X\mid Y}^{-1}\tilde{\mathbf{x}} + \tilde{\mathbf{y}}^T\left(\boldsymbol{\Sigma}_{YY}^{-1} + \mathbf{A}^T\boldsymbol{\Sigma}_{X\mid Y}^{-1}\mathbf{A}\right)\tilde{\mathbf{y}}$$

Now, let's group all terms containing $\boldsymbol{\Sigma}_{X\mid Y}^{-1}$ together:


$$Q_{\text{joint}} = \left[ \tilde{\mathbf{x}}^T \boldsymbol{\Sigma}_{X\mid Y}^{-1} \tilde{\mathbf{x}} - \tilde{\mathbf{x}}^T \boldsymbol{\Sigma}_{X\mid Y}^{-1}(\mathbf{A}\tilde{\mathbf{y}}) - (\mathbf{A}\tilde{\mathbf{y}})^T \boldsymbol{\Sigma}_{X\mid Y}^{-1}\tilde{\mathbf{x}} + (\mathbf{A}\tilde{\mathbf{y}})^T \boldsymbol{\Sigma}_{X\mid Y}^{-1}(\mathbf{A}\tilde{\mathbf{y}}) \right] + \tilde{\mathbf{y}}^T \boldsymbol{\Sigma}_{YY}^{-1}\tilde{\mathbf{y}}$$

Notice that the expression inside the brackets factors perfectly into a single quadratic form:


$$Q_{\text{joint}} = (\tilde{\mathbf{x}} - \mathbf{A}\tilde{\mathbf{y}})^T \boldsymbol{\Sigma}_{X\mid Y}^{-1}(\tilde{\mathbf{x}} - \mathbf{A}\tilde{\mathbf{y}}) + \tilde{\mathbf{y}}^T \boldsymbol{\Sigma}_{YY}^{-1}\tilde{\mathbf{y}}$$

#### Step 4: Final Substitution

Substituting back our variables $\tilde{\mathbf{x}}$, $\tilde{\mathbf{y}}$, and $\mathbf{A}$:


$$\tilde{\mathbf{x}} - \mathbf{A}\tilde{\mathbf{y}} = (\mathbf{x} - \boldsymbol{\mu}_X) - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}(\mathbf{y} - \boldsymbol{\mu}_Y) = \mathbf{x} - \boldsymbol{\mu}^*$$

Thus, the joint exponent expression simplifies cleanly to:


$$Q_{\text{joint}} = (\mathbf{x} - \boldsymbol{\mu}^*)^T \boldsymbol{\Sigma}_{X\mid Y}^{-1}(\mathbf{x} - \boldsymbol{\mu}^*) + (\mathbf{y} - \boldsymbol{\mu}_Y)^T \boldsymbol{\Sigma}_{YY}^{-1}(\mathbf{y} - \boldsymbol{\mu}_Y)$$

When computing $f_{\mathbf{X}\mid\mathbf{Y}} = \frac{f_{\mathbf{X},\mathbf{Y}}}{f_{\mathbf{Y}}}$, the exponents subtract:


$$-\frac{1}{2}Q_{\text{joint}} - \left(-\frac{1}{2}(\mathbf{y} - \boldsymbol{\mu}_Y)^T \boldsymbol{\Sigma}_{YY}^{-1}(\mathbf{y} - \boldsymbol{\mu}_Y)\right) = -\frac{1}{2}(\mathbf{x} - \boldsymbol{\mu}^*)^T \boldsymbol{\Sigma}_{X\mid Y}^{-1}(\mathbf{x} - \boldsymbol{\mu}^*)$$

Combining this exponent with the normalization ratio from Step 1 completes the direct proof.

# Q3) Continuous Fault Monitoring of a Spindle Motor

A factory monitors the health of a high-speed CNC spindle. The critical variable is the **Internal Temperature ($X$)**, but because the motor is sealed, they can only observe the **Total Spindle Power ($Y$)**.

**The Setup:**
1.  **Hidden Temperature ($X$):** The standardized temperature deviation follows a standard normal distribution:
    $$X \sim \mathscr{N}(0, 1)$$
2.  **Mechanical Load ($Q$):** The torque required to cut material creates "noise" in the power reading. It is independent of temperature and follows:
    $$Q \sim \mathscr{N}(0, 4)$$
3.  **The Measurement ($Y$):** The power sensor $Y$ measures the combined effect of thermal load and mechanical torque:
$$Y = X + Q$$

The factory manager says: *"Since $Y = X + Q$, our best estimate of temperature should just be $Y$ itself."*

###Answer:



### 1. Joint Distribution

Since $X$ and $Q$ are independent and normally distributed, any linear combination of them is also normally distributed. The observed power reading $Y = X + Q$ is a linear combination of independent Gaussians, making the vector $\mathbf{Z} = \begin{bmatrix} X \\ Y \end{bmatrix}$ **jointly multivariate normal**.

* **Show that $Y \sim \mathscr{N}(0, 5)$:**
* **Mean:** By the linearity of expectation:

$$\mu_Y = \mathbb{E}[Y] = \mathbb{E}[X + Q] = \mathbb{E}[X] + \mathbb{E}[Q] = 0 + 0 = 0$$


* **Variance:** Because $X$ and $Q$ are independent ($\text{Cov}(X,Q) = 0$):

$$\sigma_Y^2 = \text{Var}(Y) = \text{Var}(X + Q) = \text{Var}(X) + \text{Var}(Q) = 1 + 4 = 5$$



Thus, $Y \sim \mathscr{N}(0, 5)$.


* **Find the covariance $\text{Cov}(X, Y)$:**
Using the bilinear properties of covariance:

$$\text{Cov}(X, Y) = \text{Cov}(X, X + Q) = \text{Cov}(X, X) + \text{Cov}(X, Q) = \text{Var}(X) + 0 = 1$$



The full joint covariance matrix is therefore:


$$\boldsymbol{\Sigma} = \begin{bmatrix} \text{Var}(X) & \text{Cov}(X,Y) \\ \text{Cov}(Y,X) & \text{Var}(Y) \end{bmatrix} = \begin{bmatrix} 1 & 1 \\ 1 & 5 \end{bmatrix}$$

---

### 2. Information Geometry & Measurability

A random variable $X$ is $\sigma(Y)$-measurable if and only if all the information required to determine $X$ perfectly is captured by $Y$. By the **Doob-Dynkin Lemma**, this requires that $X$ can be expressed as a deterministic Borel-measurable function of $Y$ (i.e., $X = g(Y)$).

In this system, $X$ is **not** $\sigma(Y)$-measurable because the mapping from the hidden state to the observation collapses a 2D space $(X, Q)$ onto a 1D line $Y$. For any given observation $Y = y$, there are infinitely many pairs of $(x, q)$ such that $x + q = y$.

**Practical Interpretation:** If the power meter spikes to $Y = 5$, the engineer cannot know the exact state of the spindle. The reading could be caused by:

1. An extreme thermal emergency: $X = 5$ (severe overheating) and $Q = 0$ (no mechanical load).
2. A routine heavy operation: $X = 0$ (normal temperature) and $Q = 5$ (cutting a very hard material knot).

Because of this ambiguity, $X$ retains a conditional variance (uncertainty) even after $Y$ is observed.

---

### 3. Conditional Expectation (The Best Estimate)

Using the fundamental theorem for the conditional expectation of jointly Gaussian random variables:


$$\mathbb{E}[X \mid Y = y] = \mu_X + \frac{\text{Cov}(X,Y)}{\text{Var}(Y)}(y - \mu_Y)$$

Substituting our derived parameters ($\mu_X = 0$, $\mu_Y = 0$, $\text{Cov}(X,Y) = 1$, $\text{Var}(Y) = 5$):


$$\mathbb{E}[X \mid Y = y] = 0 + \frac{1}{5}(y - 0) = 0.2y$$

Expressing this as a random variable yields:


$$\widehat{X} = \mathbb{E}[X \mid Y] = 0.2Y$$

* **Signal-to-Total-Variance Ratio:**
We can rewrite the coefficient by breaking down the variance of $Y$:

$$\mathbb{E}[X \mid Y] = \left( \frac{\sigma_X^2}{\sigma_X^2 + \sigma_Q^2} \right) Y$$



This demonstrates that our optimal estimate scales the raw observation by the ratio of **useful signal variance** to **total observed variance**.

---

### 4. Verification of the Residual

Let our estimator be $\widehat{X} = 0.2Y = 0.2(X + Q)$. The residual error is defined as:


$$Z = X - \widehat{X} = X - 0.2(X + Q) = 0.8X - 0.2Q$$

* **Prove $\mathbb{E}[Z] = 0$:**

$$\mathbb{E}[Z] = \mathbb{E}[0.8X - 0.2Q] = 0.8\mathbb{E}[X] - 0.2\mathbb{E}[Q] = 0.8(0) - 0.2(0) = 0$$


* **Verify the Orthogonality Principle ($\text{Cov}(Z, Y) = 0$):**

$$\text{Cov}(Z, Y) = \text{Cov}(0.8X - 0.2Q, X + Q)$$



Expanding using the linearity of covariance:

$$\text{Cov}(Z, Y) = 0.8\text{Cov}(X,X) + 0.8\text{Cov}(X,Q) - 0.2\text{Cov}(Q,X) - 0.2\text{Cov}(Q,Q)$$



Since $\text{Cov}(X,Q) = 0$, $\text{Cov}(X,X) = 1$, and $\text{Cov}(Q,Q) = 4$:

$$\text{Cov}(Z, Y) = 0.8(1) + 0 - 0 - 0.2(4) = 0.8 - 0.8 = 0$$



Because the error $Z$ and the observation $Y$ are jointly Gaussian and uncorrelated, they are completely **independent**. This confirms that the estimator has successfully extracted all predictable information from $Y$.

### 5. Information Theory: Differential Entropy

The differential entropy of a continuous Gaussian random variable $W \sim \mathscr{N}(\mu, \sigma^2)$ is:


$$h(W) = \frac{1}{2} \ln(2\pi e \sigma^2) \text{ nats}$$

* **Differential Entropy $h(X)$:**
Since $\sigma_X^2 = 1$:

$$h(X) = \frac{1}{2} \ln(2\pi e \cdot 1) = \frac{1}{2} \ln(2\pi e) \approx 1.4189 \text{ nats}$$


* **Conditional Differential Entropy $h(X \mid Y)$:**
First, compute the conditional variance $\sigma_{X|Y}^2$:

$$\sigma_{X|Y}^2 = \sigma_X^2 - \frac{\text{Cov}(X,Y)^2}{\text{Var}(Y)} = 1 - \frac{1^2}{5} = 1 - 0.2 = 0.8$$



Because this conditional variance is constant and does not depend on the specific value of $y$, the conditional entropy is simply the entropy of the remaining error distribution:

$$h(X \mid Y) = \frac{1}{2} \ln(2\pi e \sigma_{X|Y}^2) = \frac{1}{2} \ln(2\pi e \cdot 0.8) \approx 1.3073 \text{ nats}$$



---

### 6. Mutual Information

The mutual information measures the total information shared between the hidden variable and the sensor:


$$I(X; Y) = h(X) - h(X \mid Y) = \frac{1}{2} \ln(2\pi e) - \frac{1}{2} \ln(2\pi e \cdot 0.8) = \frac{1}{2} \ln\left(\frac{1}{0.8}\right) = \frac{1}{2} \ln(1.25) \approx 0.1116 \text{ nats}$$

* **Uncertainty Reduction:**
* **In terms of entropy profile:** The absolute reduction in differential entropy relative to the initial state is $\frac{I(X;Y)}{h(X)} = \frac{0.1116}{1.4189} \approx 7.87\%$.
* **In terms of variance (standard metric for estimation):** The total variance of our uncertainty dropped from $\sigma_X^2 = 1$ to $\sigma_{X|Y}^2 = 0.8$. Observing the power meter eliminates exactly **$20\%$ of the variance-based uncertainty** about the temperature.


* **Effect of Expanding Noise to $\text{Var}(Q) = 100$:**
If the mechanical background noise dominates, the total observed variance becomes $\text{Var}(Y) = 1 + 100 = 101$.
The new conditional variance would be:

$$\sigma_{X|Y}^2 = 1 - \frac{1^2}{101} = \frac{100}{101} \approx 0.9901$$



The mutual information drops drastically:

$$I(X; Y) = \frac{1}{2} \ln\left(\frac{1}{0.9901}\right) = \frac{1}{2} \ln(1.01) \approx 0.00497 \text{ nats}$$



Under these conditions, the variance reduction drops to a negligible $0.99\%$, meaning the sensor becomes nearly useless due to the overwhelming mechanical noise.

### 7. Diagnostic Interpretation

The manager's suggestion ($\widehat{X}_{\text{naive}} = Y$) represents a naive, uncalibrated approach. While it is unconditionally unbiased on average ($\mathbb{E}[Y - X] = \mathbb{E}[Q] = 0$), it is a highly inefficient estimator because it is **conditionally biased** given a specific reading:


$$\mathbb{E}[\widehat{X}_{\text{naive}} - X \mid Y] = Y - \mathbb{E}[X \mid Y] = Y - 0.2Y = 0.8Y$$

If $Y = 5$, the manager's strategy declares that the temperature deviation is $+5$ standard deviations (an extreme, physically unlikely catastrophe).

The conditional expectation $\mathbb{E}[X \mid Y] = 0.2Y$ acts as a mathematical **shrinkage estimator**:

1. **Accounting for Probabilities:** Because a standard normal variable $X$ is tightly concentrated around $0$, an extreme reading like $Y = 5$ is statistically far more likely to be caused by a large fluctuation in the high-variance noise process ($Q \sim \mathscr{N}(0,4)$) than by an almost impossible jump in the lower-variance signal process ($X \sim \mathscr{N}(0,1)$).
2. **Preventing False Alarms:** The filter pulls (shrinks) the raw reading toward the historical baseline mean ($0$) to match this physical reality. Without this shrinkage coefficient ($\frac{1}{5}$), the factory would suffer frequent false alarms, triggering emergency shut-offs every time the cutting tool encountered ordinary hard resistance.
