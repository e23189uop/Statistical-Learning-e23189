### Q1. Statistical Mechanics of the Quantum Oscillator

**1. The Partition Function**
The canonical partition function is given by $Z(T) = \sum_{j=0}^\infty e^{-E_j / (k_B T)}$.
Let $\beta = \frac{1}{k_B T}$. Substitute the energy eigenvalues $E_j = \hbar \omega (j + \frac{1}{2})$:


$$Z(T) = \sum_{j=0}^\infty \exp\left[-\beta \hbar \omega \left(j + \frac{1}{2}\right)\right]$$


Factor out the constant term (which does not depend on $j$):


$$Z(T) = e^{-\frac{\beta \hbar \omega}{2}} \sum_{j=0}^\infty \left(e^{-\beta \hbar \omega}\right)^j$$


The sum is an infinite geometric series of the form $\sum r^j = \frac{1}{1-r}$ where $r = e^{-\beta \hbar \omega}$ (and $r < 1$ since $\beta\hbar\omega > 0$). Thus:


$$Z(T) = \frac{e^{-\frac{\hbar \omega}{2k_BT}}}{1 - e^{-\frac{\hbar \omega}{k_BT}}}$$

**2. Thermodynamic Potentials**

* **Helmholtz Free Energy ($F$):**
$F(T) = -k_B T \ln Z(T) = -k_B T \ln\left( \frac{e^{-\frac{\beta \hbar \omega}{2}}}{1 - e^{-\beta \hbar \omega}} \right)$
$F(T) = -k_B T \left[ -\frac{\hbar \omega}{2k_B T} - \ln\left(1 - e^{-\frac{\hbar \omega}{k_BT}}\right) \right] = \frac{\hbar \omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_BT}}\right)$
* **Internal Energy ($U$):**
$U(T) = -\frac{\partial \ln Z}{\partial \beta}$.
First, find $\ln Z = -\frac{\beta \hbar \omega}{2} - \ln(1 - e^{-\beta \hbar \omega})$.
$U(T) = -\left[ -\frac{\hbar \omega}{2} - \frac{1}{1 - e^{-\beta \hbar \omega}} \cdot (\hbar \omega) e^{-\beta \hbar \omega} \right] = \frac{\hbar \omega}{2} + \frac{\hbar \omega}{e^{\beta \hbar \omega} - 1}$

**3. Entropy**
$S(T) = -\left(\frac{\partial F}{\partial T}\right)$. It's easier to use the thermodynamic identity $F = U - TS \implies S = \frac{U - F}{T}$.


$$S = \frac{1}{T} \left[ \left(\frac{\hbar \omega}{2} + \frac{\hbar \omega}{e^{\beta \hbar \omega} - 1}\right) - \left(\frac{\hbar \omega}{2} + k_B T \ln(1 - e^{-\beta \hbar \omega})\right) \right]$$

$$S(T) = \frac{\hbar \omega}{T(e^{\beta \hbar \omega} - 1)} - k_B \ln(1 - e^{-\beta \hbar \omega})$$


*Verification:* $F + TS = F + (U - F) = U$. The identity is perfectly satisfied.

**4. Heat Capacity and Hyperbolic Form**


$$C_V = \frac{\partial U}{\partial T} = \frac{\partial}{\partial \beta}\left(\frac{\hbar \omega}{e^{\beta \hbar \omega} - 1}\right) \cdot \frac{\partial \beta}{\partial T}$$


Using the chain rule (where $\frac{\partial \beta}{\partial T} = -\frac{1}{k_B T^2}$):


$$C_V = \left[ -\frac{\hbar \omega \cdot \hbar \omega e^{\beta \hbar \omega}}{(e^{\beta \hbar \omega} - 1)^2} \right] \left( -\frac{1}{k_B T^2} \right) = k_B (\beta \hbar \omega)^2 \frac{e^{\beta \hbar \omega}}{(e^{\beta \hbar \omega} - 1)^2}$$


To rewrite using $\sinh(x)$: Note that $e^x - e^{-x} = 2\sinh(x)$.
Factor out $e^{\beta \hbar \omega}$ from the denominator:
$(e^{\beta \hbar \omega} - 1)^2 = e^{\beta \hbar \omega} (e^{\frac{\beta \hbar \omega}{2}} - e^{-\frac{\beta \hbar \omega}{2}})^2 = e^{\beta \hbar \omega} \left[ 2 \sinh\left(\frac{\beta \hbar \omega}{2}\right) \right]^2 = 4 e^{\beta \hbar \omega} \sinh^2\left(\frac{\beta \hbar \omega}{2}\right)$.
Substitute this back:


$$C_V = k_B (\beta \hbar \omega)^2 \frac{e^{\beta \hbar \omega}}{4 e^{\beta \hbar \omega} \sinh^2(\frac{\beta \hbar \omega}{2})} = k_B \left(\frac{\hbar \omega}{2k_BT}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar \omega}{2k_BT}\right)}$$

**5. Asymptotic Behavior (Discussion)**
In the high-temperature limit ($k_B T \gg \hbar \omega$), $x = \frac{\hbar \omega}{2k_B T} \to 0$.
For small $x$, the Taylor expansion of $\sinh(x) \approx x$.
Therefore, $C_V \approx k_B (x)^2 \frac{1}{x^2} = k_B$.
**Yes, it matches the classical equipartition theorem.** A 1D harmonic oscillator has two degrees of freedom (kinetic and potential energy), each contributing $\frac{1}{2} k_B$ to the heat capacity, yielding a total of $k_B$.

### Q2. The Maximum Entropy Distribution under Mean and Variance

To find the probability density $p(x)$ that maximizes entropy $H = -\int p(x) \ln p(x) dx$ subject to constraints, use the method of Lagrange multipliers. The constraints are:

1. $\int p(x) dx = 1$ (Normalization)
2. $\int x p(x) dx = \mu$ (Mean)
3. $\int (x-\mu)^2 p(x) dx = \sigma^2$ (Variance)

Set up the Lagrangian functional $L$:


$$L = -\int p(x) \ln p(x) dx + \lambda_0 \left(\int p(x) dx - 1\right) + \lambda_1 \left(\int x p(x) dx - \mu\right) + \lambda_2 \left(\int (x-\mu)^2 p(x) dx - \sigma^2\right)$$


Taking the functional derivative with respect to $p(x)$ and setting it to zero:


$$\frac{\delta L}{\delta p(x)} = -\ln p(x) - 1 + \lambda_0 + \lambda_1 x + \lambda_2 (x - \mu)^2 = 0$$


Solving for $p(x)$:


$$p(x) = \exp\left(\lambda_0 - 1 + \lambda_1 x + \lambda_2 (x - \mu)^2\right)$$


Because the variance constraint represents dispersion around the mean symmetrically, the linear term $\lambda_1 x$ drops out (i.e., $\lambda_1 = 0$) to keep the peak at $\mu$. Thus:


$$p(x) = C \exp(\lambda_2 (x-\mu)^2)$$


For $p(x)$ to be a valid probability distribution that converges, $\lambda_2$ must be negative. Let $\lambda_2 = -\frac{1}{2\sigma^2}$ to satisfy the variance constraint, and the normalization constant $C = \frac{1}{\sqrt{2\pi\sigma^2}}$ to satisfy the area under the curve.
**Result:**

$$X \sim \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$


This proves the Normal (Gaussian) distribution is the maximum entropy distribution for a given mean and variance.

### Q3. Multiclass Logistic Regression: Automatic Classification of Emails

Note: No explicit question asked, formulated the MaxEnt probability distributions for the given constraints.

In a Maximum Entropy classifier (which is mathematically identical to Logistic Regression), the probability of routing an email to target $Y=c$ given input features $\mathbf{x}$ is modeled as:


$$P(Y = c \mid \mathbf{x}) = \frac{\exp(\mathbf{w}_c \cdot \mathbf{x})}{Z(\mathbf{x})}$$


where $Z(\mathbf{x}) = \sum_{y \in Y} \exp(\mathbf{w}_y \cdot \mathbf{x})$ and $\mathbf{w}_c$ is the weight vector to be learned.
Based on the provided constraints, the optimization algorithm will adjust the weights $w_{1, Y=1}, w_{2, Y=1}, w_{3, Y=1}$ such that the predicted expected values exactly match the empirical data constraints:

1. $\sum_{\mathbf{x}} P(\mathbf{x}) \cdot x_1 \cdot P(Y=1 \mid \mathbf{x}) = 0.05$ (Pushes $w_{1, Y=1}$ to be highly negative).
2. $\sum_{\mathbf{x}} P(\mathbf{x}) \cdot x_2 \cdot P(Y=1 \mid \mathbf{x}) = 0.15$ (Pushes $w_{2, Y=1}$ to be neutral/positive).

### Q4. Asset Pricing and Portfolio Optimization

**Objective:** Formulate the Maximum Entropy problem to find $p(S_T)$.

We want to maximize the Shannon entropy $H = -\int_0^\infty p(S_T) \ln p(S_T) dS_T$ subject to four continuous constraints provided by the market:

1. **Normalization:** $\int_0^\infty p(S_T) dS_T = 1$
2. **Expected Price:** $\int_0^\infty S_T p(S_T) dS_T = 102$
3. **Expected Variance/Squared Price:** $\int_0^\infty S_T^2 p(S_T) dS_T = 10600$
4. **Logarithmic/Crash Risk:** $\int_0^\infty \ln(S_T) p(S_T) dS_T = 4.60$

**Formulation:**
Using Lagrange multipliers $(\lambda_0, \lambda_1, \lambda_2, \lambda_3)$ for each constraint respectively, we write the Lagrangian and set its variation to zero, leading to the exponentiated form:


$$p(S_T) = \exp\left( -1 - \lambda_0 - \lambda_1 S_T - \lambda_2 S_T^2 - \lambda_3 \ln(S_T) \right)$$


Using exponent rules, we can pull the logarithmic term down as a polynomial base:


$$p(S_T) = \exp(-1-\lambda_0) \cdot \exp(-\lambda_3 \ln(S_T)) \cdot \exp(-\lambda_1 S_T - \lambda_2 S_T^2)$$

$$p(S_T) = \frac{1}{Z} S_T^{-\lambda_3} \exp\left(-\lambda_1 S_T - \lambda_2 S_T^2\right)$$


*(where $Z = \exp(1+\lambda_0)$ is the normalizing partition function).*

**Solution Interpretation:** The resulting distribution is a **hybrid between a Lognormal and a Truncated Gaussian distribution** (often related to the Generalized Gamma distribution family). The $-\lambda_3$ term creates the "fat tail" and skew observed in financial markets, protecting your portfolio optimization from the flawed assumption of a perfect bell curve.
